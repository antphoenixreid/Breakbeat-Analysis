```python
import yt_dlp
import ffmpeg
import numpy as np
import librosa
import soundfile as sf
import io
```

```python
def search_youtube(song_name, artist_name=None):
    query = f"{song_name} {artist_name} official audio" if artist_name else song_name
    search_url = f"ytsearch: {query}"

    ydl_opts = {
        "format": "bestaudio/best",
        "quiet": True
    }

    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        result = ydl.extract_info(search_url, download=False)

    if "entries" in result:
        return result["entries"][0]["url"] # Get the first result
    
    return None
```

```python
# Test Use Case of search_youtube
song_url = search_youtube("Squabble Up", "Kendrick Lamar")
print("Best Match:", song_url)
```

```python
def stream_youtube_audio(url):
    process = (
        ffmpeg.input(url)
        .output('pipe:', format='wav', acodec='pcm_s16le', ac=1, ar=44100)
        .run_async(pipe_stdout=True, pipe_stderr=True, quiet=True)
    )

    audio_data, _ = process.communicate()
    audio_buffer = io.BytesIO(audio_data)

    # Read as WAV file and process with librosa
    y, sr = sf.read(audio_buffer, dtype='float32')

    return y, sr
```

```python
if song_url:
    y, sr = stream_youtube_audio(song_url)
```
